#UCI wine_list

```python
from sklearn.datasets import load_wine
import numpy as np
# Load the real UCI Wine public dataset
wine = load_wine()
# Extract the raw data into a 2D NumPy array of shape (178, 13)
data = wine.data
feature_names = wine.feature_names
print("Dataset loaded successfully!")
print("Feature columns:", feature_names)
```

```python
numpy_array = np.array(data)
print(numpy_array.dtype)
print(f"Array shape: {numpy_array.shape}")
print(f"N of dimensions: {numpy_array.ndim}")
```

```python
print(f"First 5 wines:")
np.set_printoptions(suppress=True, precision=2)
print(numpy_array[:5,:])
```

```python
print(f"Alcohol content:")
print(numpy_array[:10,0])
```

```python
print(numpy_array[10:21,:3])
```

```python
print(f"Color intensity:\n{numpy_array[-5:,9]}")
print(f"Proline:\n{numpy_array[-5:,12]}")
```

```python
#Mean of each feature
for i in range(13):
  print(f"{feature_names[i].upper()} mean: {np.mean(numpy_array[:,i])}\n")
```

```python
print(f"Highest alcohol wine: {np.max(numpy_array[:,0])}")
print(f"Minimun alcohol wine: {np.min(numpy_array[:,0])}")
```

```python
high_alcohol_grade = numpy_array[numpy_array[:,0] > 14]
np.set_printoptions(suppress=True, precision=2)
print("High alcohol wines:")
print(len(high_alcohol_grade))
print(high_alcohol_grade)
```

```python
print("Scaled alcohol values:")
(high_alcohol_grade[:,0] - np.min(high_alcohol_grade[:,0])) / (np.max(high_alcohol_grade[:,0]) - np.min(high_alcohol_grade[:,0]))
```
